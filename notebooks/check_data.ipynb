## Database Configuration

## Connect to Postgres

```python
conn = psycopg2.connect(**DB_CONFIG)
```

## Run Queries to Verify the Data

```python
sql = "SELECT * FROM reviews;"

run_query(sql, conn)
```

## Count reviews per bank

```python
sql = """
SELECT r.bank_name, COUNT(*) AS total_reviews
FROM reviews v
JOIN banks r ON r.bank_id = v.bank_id
GROUP BY r.bank_name;
"""

run_query(sql, conn)
```

## Average rating per bank

```python
sql = """
SELECT r.bank_name, ROUND(AVG(v.rating), 2) AS avg_rating
FROM reviews v
JOIN banks r ON r.bank_id = v.bank_id
GROUP BY r.bank_name;
"""

run_query(sql, conn)
```

## Check for nulls

```python
sql = """
SELECT *
FROM reviews
WHERE
    review_text IS NULL
    OR rating IS NULL
    OR review_date IS NULL
    OR sentiment_label IS NULL
    OR sentiment_score IS NULL
    OR identified_theme IS NULL
    OR bank_id IS NULL;
"""

run_query(sql, conn)
```

> No null values in any of the columns and average rating, total review per bank and total rows consistent with CSV.
